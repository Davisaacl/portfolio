```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv")
```

Validate the type of information in the dataframe

```python
# Start coding here! Use as many cells as you like

netflix_df.head(20)
```

Let's look for only the movies that were released during 1990s

```python
netflix_df_1990s = netflix_df[(netflix_df['release_year']>=1990) & (netflix_df['release_year']<=1999)]

# For knowing the number of columns and rows
print(netflix_df_1990s.shape)

netflix_df_1990s.head()
```

Now that I have a dataframe with all the movies from 1990 to 1999, let's consider the duration of all the movies.

```python
# Create a sub dataframe with only the movie durations.
durations_df = netflix_df_1990s[['duration']].reset_index(drop=True)
durations_df.shape
```

Calulate the most frequent movie duration in the 1990s.

```python
duration_mode = durations_df.mode()
duration = int(duration_mode.iloc[0,0])
# type(int(duration))
duration
```

I can conclude that the most common duration is 94, as it is the most repeated duration.

Now, let's study when an action movie is considered short. For this part, I know that short short movies have a duration less than 90 minutes, and I consider only 'action' movies.

```python
# Short action movies from the 1900s
short_movies_action_df = netflix_df_1990s[(netflix_df_1990s['duration'] < 90) & (netflix_df_1990s['genre'] =='Action')]
short_movies_action_df
```

Counting the number of rows gives us the number of movies that I am looking for.

```python
short_movie_count = short_movies_action_df.shape[0]
print(short_movie_count)
```

In this Netflix catalogue, there are only 7 Action movies with a duration less than 90 minutes, realeased during 1990s.
